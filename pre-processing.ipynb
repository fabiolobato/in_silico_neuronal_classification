### Rename columns

```python
##Removing whitespace in each column name
df = df.rename(columns=lambda x: x.replace(" ", "_"))
```

### Converting string to lowercase and removing accents

```python
import unicodedata
from unicodedata import normalize
import re
```

```python
##Defining function that converts to lowercase and removes accents
def pre_processing(text):
    ##text adaptation 
    adp_text = str(text).lower()
    acent_text = normalize('NFKD', adp_text).encode('ASCII','ignore').decode('ASCII')
    return acent_text
```

```python
##Apply function all Dataframe
df = df.applymap(pre_processing)
```

```python
##Remove spaces after commas
df = df.replace(', ', ',', regex=True)
df = df.replace(' ,', ',', regex=True)
df = df.replace(' ','_', regex=True)
```
